请安装并导入外部包：numpy、pandas，分别赋予别名：np和pd；  
1. 读取data.csv文件中的数据；   
2. 计算两个变量之间的皮尔逊相关（注：可以用第三方包scipy里的pearsonr）   


```python
# 导入 numpy 和pandas 库, 约定俗成别名为 np和pd
import numpy as np
import pandas as pd
from scipy.stats import pearsonr    

raw_data = pd.read_csv("data.csv")
print(raw_data)

#计算皮尔逊相关
age = raw_data['Age']
weight = raw_data['Weight']
age_weight_corr = np.corrcoef(age, weight)
print(age_weight_corr)
print(pearsonr(age, weight))
```

#### Pandas数据抽取

```python
import pandas as pd
data = [[110,105,99],[105,88,115],[109,120,130],[112,115]]
name = ['张三','李四','赵五','王六']
columns = ['语文','数学','英语']
df = pd.DataFrame(data=data, index=name, columns=columns)
print(df)

# 抽取一行数据
print(df.loc['张三'])
print(df.iloc[0])

# 抽取多行数据
print(df.loc[['张三', '赵五']])
print(df.iloc[[0,2]])

# 抽取连续任意多行数据
print(df.loc['张三':'赵五'])
print(df.iloc[0:4])
```

```python
import pandas as pd
data = [[110,105,99],[105,88,115],[109,120,130],[112,115]]
name = ['张三','李四','赵五','王六']
columns = ['语文','数学','英语']
df = pd.DataFrame(data=data, index=name, columns=columns)
print(df)

# 抽取指定列数据，直接使用列名
print(df.数学)
print(df['数学'])
print(df[['语文', '数学']]) 

# loc属性和iloc属性都包含了两个参数，第一个参数代表行
# 第二个参数代表列，那么在抽取指定列数据时，行参数不能省略。
print(df.loc[:, ['语文', '数学']]) 
print(df.iloc[:, [0,1]]) 
```

```python
import pandas as pd
data = [[110,105,99],[105,88,115],[109,120,130],[112,115]]
name = ['张三','李四','赵五','王六']
columns = ['语文','数学','英语']
df = pd.DataFrame(data=data, index=name, columns=columns)

#抽取指定行列数据
print(df.loc['张三','英语'])              #“英语”成绩 输出一个数字，不是dataframe
print(df.loc[['张三'],['英语']])          #“张三”的“英语”成绩
print(df.loc[['张三'],['数学','英语']])   #“张三”的“数学”和“英语”成绩
print(df.iloc[[1],[2]])                       #第2行第3列
print(df.iloc[1:,[2]])                        #第2行到最后一行的第3列
print(df.iloc[1:,[0,2]])                      #第2行到最后一行的第1列和第3列
print(df.iloc[:,2])                           #所有行，第3列
```

#### Pandas数据计算

```python
import pandas as pd
data = [[110,105,99],[105,88,115],[109,120, ]]
index = [1,2,3]
columns = ['语文','数学','英语']
df = pd.DataFrame(data=data, index=index, columns=columns)
print(df)

print(df.sum())

df['总成绩'] = df.sum(axis=1, skipna=False)  #增加总成绩这一列
print(df)

df.loc[:,'总成绩'] = df.sum(axis=1)  #增加总成绩这一列
print(df)
```

2. 某图书馆记录了周一至周五三类图书的借阅量，单位为“本”。  
| 星期 | 文学 | 科技 | 历史 |  
| 周一 | 120 | 80 | 50 |  
| 周二 | 100 | 90 | 55 |  
| 周三 | 140 | 75 | 60 |  
| 周四 | 110 | 85 | 45 |  
| 周五 | 160 | 100 | 70 |  
请使用Pandas完成以下任务：  
a）分别使用**列表**和**字典**创建两个内容相同的DataFrame。均以星期为行索引，以“文学”“科技”“历史”为列名。      
b) 使用`loc`抽取周二至周四的“文学”和“科技”借阅量，再使用`iloc`完成相同的抽取。说明两种方法在索引含义和切片结束位置上的区别。    
c) 使用`sum()`分别计算每类图书五天的借阅总量，以及每天三类图书的借阅总量。注明两次计算使用的`axis`值。  
d) 使用`mean()`计算每类图书的日均借阅量，并说明返回结果的类型。  


```python
import pandas as pd

# 1. 使用列表创建DataFrame
weekdays = ['周一', '周二', '周三', '周四', '周五']
categories = ['文学', '科技', '历史']

data_list = [
    [120, 80, 50],
    [100, 90, 55],
    [140, 75, 60],
    [110, 85, 45],
    [160, 100, 70]
]

df_list = pd.DataFrame(
    data_list,
    index=weekdays,
    columns=categories
)
print(df_list)

# 使用字典创建内容相同的DataFrame
data_dict = {
    '文学': [120, 100, 140, 110, 160],
    '科技': [80, 90, 75, 85, 100],
    '历史': [50, 55, 60, 45, 70]
}

df_dict = pd.DataFrame(data_dict, index=weekdays)
print(df_dict)

# 2. 使用loc和iloc抽取相同的数据
result_loc = df_list.loc['周二':'周四', ['文学', '科技']]
result_iloc = df_list.iloc[1:4, 0:2]

print(result_loc)
print(result_iloc)

# 3. 计算每类图书和每天的借阅总量
category_total = df_list.sum(axis=0)
daily_total = df_list.sum(axis=1)

print(category_total)
print(daily_total)

# 4. 计算每类图书的日均借阅量
category_mean = df_list.mean(axis=0)

print(category_mean)
print(type(category_mean))
```

3. 练习二数据文件夹包含三个文件：study_hours.xlsx、study_hours.csv和study_hours.txt。  
它们保存的是同一份模拟数据，记录了5名学生三天的课外学习时长，单位为“小时”。  
文件中的缺失数据以空单元格或空字段表示。  
0表示当天没有进行课外学习，缺失表示当天的学习时长未记录。  
请使用Pandas完成以下任务：  
a) 分别导入Excel、CSV和TXT文件，保存为三个DataFrame，并输出各自的内容及行列数量。CSV使用逗号分隔，TXT使用制表符分隔。  
b) 后续分析使用从Excel导入的DataFrame。  
c) 使用loc抽取“周一”至“周三”三列，保存为hours，确保其中不包含姓名列。使用iloc抽取hours中前3名学生在周一和周二的学习时长。  
d) 使用sum()和mean()，设置skipna=True，计算每名学生已记录的学习总时长及有记录日期的平均学习时长。  

注意：保留原始缺失值，不进行填补或删除。

```python
import pandas as pd

# 1. 分别导入Excel、CSV和TXT文件
df_excel = pd.read_excel(r'exercise02_data\study_hours.xlsx')

df_csv = pd.read_csv(r'exercise02_data\study_hours.csv')

df_txt = pd.read_csv(r'exercise02_data\study_hours.txt',sep='\t')

print(df_excel)
print(df_excel.shape)

print(df_csv)
print(df_csv.shape)

print(df_txt)
print(df_txt.shape)

# 2. 使用loc选取三天的学习时长，不包含姓名列
hours = df_excel.loc[:, '周一':'周三']
print(hours)

# 3. 抽取前三名学生在周一和周二的学习时长
part = hours.iloc[0:3, 0:2]
print(part)

# 4. 按学生计算，跳过缺失值
student_total = hours.sum(axis=1, skipna=True)
student_mean = hours.mean(axis=1, skipna=True)

print(student_total)
print(student_mean)

```
